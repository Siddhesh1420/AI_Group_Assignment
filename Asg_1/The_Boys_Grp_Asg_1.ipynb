```python
# all the needed libiray for this
import heapq as pq
import time
```

```python
# input function that take the graph as input
def input_graph() -> tuple[int, int, dict[str, list[tuple[str, int]]], dict[str,int], str, str]:
    adjList = {}
    N = int()
    M = int()
    heuristic = {}

    while(True):
        try:
            firstLine = input().split()

            if(len(firstLine) == 2):
                N = int(firstLine[0])
                M = int(firstLine[1])

                if 2 <= N <= 100:
                    if 1 <= M <= ((N * (N-1)) / 2):
                        break

            print("Invalid Input!")
            continue

        except:
            print("Invalid Input!")
            continue

    for _ in range(M):
        while(True):
            try:
                mLine = input().split()

                if(len(mLine) == 3):
                    U = mLine[0]
                    V = mLine[1]
                    C = int(mLine[2])

                    if C >= 0:
                        adjList[U] = adjList.get(U,[]) + [(V,C)]
                        adjList[V] = adjList.get(V,[]) + [(U,C)]
                        break

                print("Invalid Input!")
                continue

            except:
                print("Invalid Input!")
                continue 

    S = str()
    G = str()

    while(True):
        try:
            secondLine = input().split()

            if(len(secondLine) == 2):
                S = secondLine[0]
                G = secondLine[1]
                break

            print("Invalid Input!")
            continue

        except:
            print("Invalid Input!")
            continue

    for _ in range(N):
        while(True):
            try:
                thirdLine = input().split()

                if(len(thirdLine) == 2):
                    U = thirdLine[0]
                    H = int(thirdLine[1])
                    heuristic[U] = H
                    break

                print("Invalid Input!")
                continue

            except:
                print("Invalid Input!")
                continue

    return N, M, adjList, heuristic, S, G
```

```python
# print the input in better way
def print_input(N : int, M : int, adjList : dict[str, list[tuple[str, int]]], heuristic : dict[str, int], S : str, G : str) -> None:
    print(f"{N} {M}")
    visted = set()
    for node in adjList:
        for neg, cost in adjList[node]:
            if neg not in visted:
                print(f"{node} {neg} {cost}")
        visted.add(node)
    print(f"{S} {G}")
    for node in heuristic:
        print(f"{node} {heuristic[node]}")
    print("\n")
    return
```

```python
# implemation of the greed best firsrt search
def greedy_best_first_search(adjList : dict[str, list[tuple[str, int]]], heuristic : dict[str, int], S : str, G : str) -> tuple[list[str] | None, int | None, int]:
    minHeap = []
    nodesExpanded = 0
    pq.heappush(minHeap,(heuristic[S],S,[S],0))
    while len(minHeap) > 0:
        _, curr, path, totalCost = pq.heappop(minHeap)
        nodesExpanded += 1
        if curr == G:
            return path, totalCost, nodesExpanded
            pass
        for neighbour,cost in adjList[curr]:
            pq.heappush(minHeap,(heuristic[neighbour],neighbour,path + [neighbour],cost + totalCost))
    return None, None, nodesExpanded
```

```python
# implemnation of the a* search
def a_star_search(adjList : dict[str, list[tuple[str, int]]], heuristic : dict[str, int], S : str, G : str) -> tuple[list[str] | None, int | None, int]:
    minHeap = []
    nodesExpanded = 0
    pq.heappush(minHeap,(heuristic[S],S,[S],0))
    while len(minHeap) > 0:
        _, curr, path, totalCost = pq.heappop(minHeap)
        nodesExpanded += 1
        if curr == G:
            return path, totalCost, nodesExpanded
            pass
        for neighbour,cost in adjList[curr]:
            pq.heappush(minHeap,(totalCost+cost+heuristic[neighbour],neighbour,path + [neighbour],cost + totalCost))
    return None, None, nodesExpanded
```

```python
# print the expected output
def print_result(path : list[str], totalCost : int,nodesExpanded : int, algo : str,time : float) -> None:
    print("Algorithm:",algo)
    if path is None:
        print("Path Found: No")
        print("Nodes Expanded =",nodesExpanded)
        print(f"Execution Time = {time:.8f}s")
        return
    print("Path Found: Yes")
    print("Path:"," -> ".join(path))
    print("Total Cost =",totalCost)
    print("Nodes Expanded =",nodesExpanded)
    print(f"Execution Time = {time:.8f}s")
    return
```

```python
# function that do all the things together
def main() -> None:
    N, M, adjList, heuristic, S, G = input_graph()
    print_input(N, M, adjList, heuristic, S, G)
    start_time = time.perf_counter()
    path, totalCost, nodesExpanded = greedy_best_first_search(adjList, heuristic, S, G)
    end_time = time.perf_counter()
    print_result(path,totalCost,nodesExpanded,"Greedy Best-First Search",end_time - start_time)
    start_time = time.perf_counter()
    path, totalCost, nodesExpanded = a_star_search(adjList, heuristic, S, G)
    end_time = time.perf_counter()
    print("\n")
    print_result(path,totalCost,nodesExpanded,"A* Search",end_time-start_time)
    return
```

```python
# function that will run on the hardcode inputs
def sample_input(inputTuple : tuple[int, int, dict[str, list[tuple[str, int]]], dict[str, int], str, str, int]) -> None:
    N, M, adjList, heuristic, S, G, i = inputTuple
    print(f"Sample Input {i}")
    print_input(N, M, adjList, heuristic, S, G)
    start_time = time.perf_counter()
    path, totalCost, nodesExpanded = greedy_best_first_search(adjList, heuristic, S, G)
    end_time = time.perf_counter()
    print(f"Sample Output {i}")
    print_result(path,totalCost,nodesExpanded,"Greedy Best-First Search",end_time - start_time)
    start_time = time.perf_counter()
    path, totalCost, nodesExpanded = a_star_search(adjList, heuristic, S, G)
    end_time = time.perf_counter()
    print("\n")
    print_result(path,totalCost,nodesExpanded,"A* Search",end_time-start_time)
    return
```

```python
R, C = 5, 7

board = [
    list("A.E.#.."),
    list(".#..#E."),
    list("...#..."),
    list("#E#..#."),
    list("....E.B")
]

starting_player = "MAX"

depth = 5
```

```python
A_position=None
B_position=None

# We use a set as it is more efficient to check if a cell is in the set or not.
remaining_energy_cells=set() 

for i in range(R):
    for j in range(C):
        if board[i][j]=='A':
            A_position=(i,j)
        elif board[i][j]=='B':
            B_position=(i,j)
        elif board[i][j]=='E':
            remaining_energy_cells.add((i,j))
```

```python
max_score=0
min_score=0
current_player=starting_player

initial_state=(A_position,B_position,remaining_energy_cells,current_player,max_score,min_score)
```

```python
moves={"UP":(-1,0),"RIGHT":(0,1),"DOWN":(1,0),"LEFT":(0,-1)}

def get_new_position(position,move):
    new_row=position[0]+moves[move][0]
    new_col=position[1]+moves[move][1]
    if 0<=new_row<R and 0<=new_col<C:
        return (new_row,new_col)
    else:
        return None
    
def is_valid_move(new_pos,other_robot_pos):
    if new_pos is None:
        return False
    if new_pos==other_robot_pos:
        return False
    if board[new_pos[0]][new_pos[1]]=='#':
        return False
    return True
```

```python
def successor(state,move):
    A_pos,B_pos,remaining_energy_cells,current_player,max_score,min_score=state
    if current_player=='MAX':
        new_A_pos=get_new_position(A_pos,move)
        if not is_valid_move(new_A_pos,B_pos):
            return None
        new_remaining_energy_cells=remaining_energy_cells.copy()
        if new_A_pos in new_remaining_energy_cells:
            new_remaining_energy_cells.remove(new_A_pos)
            max_score+=10
        new_state=(new_A_pos,B_pos,new_remaining_energy_cells,'MIN',max_score,min_score)
        return new_state
    elif current_player=='MIN':
            new_B_pos=get_new_position(B_pos,move)
            if not is_valid_move(new_B_pos,A_pos):
                return None
            new_remaining_energy_cells=remaining_energy_cells.copy()
            if new_B_pos in new_remaining_energy_cells:
                new_remaining_energy_cells.remove(new_B_pos)
                min_score+=10
            new_state=(A_pos,new_B_pos,new_remaining_energy_cells,'MAX',max_score,min_score)
            return new_state
```

```python
def get_successors(state,stats):
    successors=[]
    for move in moves.keys():
        new_state=successor(state,move)
        if new_state is not None:
            successors.append((move,new_state))
            stats['nodes_generated']+=1
    return successors
```

```python
def manhattan_distance(pos1,pos2):
    return abs(pos1[0]-pos2[0])+abs(pos1[1]-pos2[1])

def positional_advantage(A_pos,B_pos,remaining_energy_cells):
    if not remaining_energy_cells:
        return 0
    A_distances=[manhattan_distance(A_pos,cell) for cell in remaining_energy_cells]
    B_distances=[manhattan_distance(B_pos,cell) for cell in remaining_energy_cells]
    return min(B_distances)-min(A_distances)
```

```python
def evaluation_function(state):
    A_pos,B_pos,remaining_energy_cells,current_player,max_score,min_score=state
    return (max_score-min_score)+positional_advantage(A_pos,B_pos,remaining_energy_cells)
```

```python
def minimax(state,depth,stats):
    A_pos,B_pos,remaining_energy_cells,current_player,max_score,min_score=state
    if depth==0 or not remaining_energy_cells:
        return evaluation_function(state)
    stats['nodes_expanded']+=1
    
    successors=get_successors(state,stats)
    
    if current_player=='MAX':
        best_score=float('-inf')
        for move,new_state in successors:
            score=minimax(new_state,depth-1,stats)
            best_score=max(best_score,score)
        return best_score
    
    else:
        best_score=float('inf')
        for move,new_state in successors:
            score=minimax(new_state,depth-1,stats)
            best_score=min(best_score,score)
        return best_score
    
```

### Cost function

```python
def n_conflict(state , course_pairs):

  count = 0

  for (c1,c2) in course_pairs:
    if state[c1-1] == state[c2-1]:
      count += 1

  return count


def n_excess_courses( state , S ):

    tot_count = 0

    for i in range(S):
      count = 0
      for slot in state:
        if slot == i+1 :
          count=count+1
      tot_count += max(0, count - 2)

    return tot_count

def calculate_cost(state , conflicts , S ):

    n_course_conf = n_conflict(state, conflicts)

    conf_penalty = n_course_conf * 10

    n_exc_courses = n_excess_courses(state , S)

    dist_penalty = n_exc_courses * 2

    return conf_penalty , dist_penalty , conf_penalty + dist_penalty
```

### Neighbor Generation

```python
def generate_neighbors(initial_state , S):

  all_neighbours = []

  for i in range(len(initial_state)):
    for j in range(S):
      if initial_state[i] == j+1 :
        continue

      new_neighbour = initial_state.copy()
      new_neighbour[i] = j+1

      all_neighbours.append(new_neighbour)

  return all_neighbours
```

### Hill Climbing

```python
def hill_climbing(initial_state , conflicts ,S):

  x , y , initial_state_cost = calculate_cost(initial_state , conflicts , S)

  all_neighbours = generate_neighbors(initial_state , S)

  best_neighbour = initial_state
  best_neighbour_cost = initial_state_cost

  for neighbour in all_neighbours:

    x , y , neighbour_cost = calculate_cost(neighbour , conflicts , S)

    if neighbour_cost < best_neighbour_cost:
      best_neighbour = neighbour
      best_neighbour_cost = neighbour_cost

  return best_neighbour , best_neighbour_cost
```

### Main Function

```python
import time as time

def main(N,S,K,conflicts,initial_state):

  start_time = time.time()
  print("========================================")
  print("Algorithm: Hill Climbing")
  print("Problem: Exam Timetable Optimization")
  print("========================================")

  print("Initial Timetable :")

  for i in range(N):
    print("C",(i+1),"-> Slot " ,initial_state[i])

  conflict_cost , dist_cost , total_cost = calculate_cost(initial_state , conflicts , S)
  print("Initial Conflict Cost = ",conflict_cost)
  print("Initial Distribution Cost = ",dist_cost)
  print("Initial Total Cost = ",total_cost)

  i=1

  while(1):

    best_neighbour,best_neighbour_cost = hill_climbing(initial_state , conflicts , S)

    print("----------------------------------------")
    print("Iteration ",i)
    print("Best Neighbor Timetable:")
    for j in range(N):
      print("C",(j+1),"-> Slot " ,best_neighbour[j])

    neighbour_conflict_cost , neighbour_dist_cost , neighbour_total_cost = calculate_cost(best_neighbour , conflicts , S)
    print("Conflict Cost = ",neighbour_conflict_cost)
    print("Distribution Cost = ",neighbour_dist_cost)
    print("Total Cost = ",neighbour_total_cost)

    if best_neighbour_cost >= total_cost:
      break

    initial_state = best_neighbour
    total_cost = best_neighbour_cost
    i=i+1

  print("========================================")
  print("Final Timetable :")

  for j in range(N):
    print("C",(j+1),"-> Slot " ,initial_state[j])

  conflict_cost , dist_cost , total_cost = calculate_cost(initial_state , conflicts , S)
  print("Final Conflict Cost = ",conflict_cost)
  print("Final Distribution Cost = ",dist_cost)
  print("Final Total Cost = ",total_cost)

  print("Interations = ",i)
  print("Total Neighboring States Evaluated = ",(18*i))
  end_time = time.time()
  print("Execution Time = ",end_time-start_time," seconds")

  if total_cost == 0:
    print("Termination Reason: Global Optimum (Ideal Solution)")
  else:
    print("Termination Reason: Local Optimum")
  print("========================================")
```
