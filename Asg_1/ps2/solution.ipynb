```python
import heapq as pq
```

```python
# N is number of node
# M is number of edges
# S start point
# G goal point
```

```python
def inputGraph():
    adjList = {}
    N = int()
    M = int()
    heuristic = {}

    while(True):
        try:
            firstLine = input().split()

            if(len(firstLine) == 2):
                N = int(firstLine[0])
                M = int(firstLine[1])

                if 2 <= N <= 100:
                    if 1 <= M <= ((N * (N-1)) / 2):
                        break

            print("Invalid Input!")
            continue

        except:
            print("Invalid Input!")
            continue

    for _ in range(M):
        while(True):
            try:
                mLine = input().split()

                if(len(mLine) == 3):
                    U = mLine[0]
                    V = mLine[1]
                    C = int(mLine[2])

                    if C >= 0:
                        adjList[U] = adjList.get(U,[]) + [(V,C)]
                        adjList[V] = adjList.get(V,[]) + [(U,C)]
                        break

                print("Invalid Input!")
                continue

            except:
                print("Invalid Input!")
                continue 

    S = str()
    G = str()

    while(True):
        try:
            secondLine = input().split()

            if(len(secondLine) == 2):
                S = secondLine[0]
                G = secondLine[1]
                break

            print("Invalid Input!")
            continue

        except:
            print("Invalid Input!")
            continue

    for _ in range(N):
        while(True):
            try:
                thirdLine = input().split()

                if(len(thirdLine) == 2):
                    U = thirdLine[0]
                    H = int(thirdLine[1])
                    heuristic[U] = H
                    break

                print("Invalid Input!")
                continue

            except:
                print("Invalid Input!")
                continue

    return N, M, adjList, heuristic, S, G
```

```python
def greedyBestFirstSearch(adjList, heuristic, S, G):
    minHeap = []
    nodesExpanded = 1
    pq.heappush(minHeap,(heuristic[S],S,[S],0))
    while len(minHeap) > 0:
        _, curr, path, totalCost = pq.heappop(minHeap)
        nodesExpanded += 1
        if curr == G:
            return path, totalCost, nodesExpanded
            pass
        for neighbour,cost in adjList[curr]:
            pq.heappush(minHeap,(heuristic[neighbour],neighbour,path + [neighbour],cost + totalCost))
    return None, None, nodesExpanded
```

```python
def printResult(path, totalCost,nodesExpanded, algo):
    print("Algorithm :",algo)
    if path is None:
        print("Path Found: No")
        print("Nodes Expanded =",nodesExpanded)
        return
    print("Path Found: Yes")
    print("Path:"," -> ".join(path))
    print("Total Cost =",totalCost)
    print("Nodes Expanded =",nodesExpanded)
```

```python
def main():
    N, M, adjList, heuristic, S, G = inputGraph()
    path, totalCost, nodesExpanded = greedyBestFirstSearch(adjList, heuristic, S, G)
    printResult(path,totalCost,nodesExpanded,"Grredy BFS")
```

```python
if __name__ == "__main__":
    main()
```
