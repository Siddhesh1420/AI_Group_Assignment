# Cost Function

```python
def n_conflict(state , course_pairs):

  count = 0

  for (c1,c2) in course_pairs:
    if state[c1-1] == state[c2-1]:
      count += 1

  return count


def n_excess_courses( state , S ):

    tot_count = 0

    for i in range(S):
      count = 0
      for slot in state:
        if slot == i+1 :
          count=count+1
      tot_count += max(0, count - 2)

    return tot_count

def calculate_cost(state , conflicts , S ):

    n_course_conf = n_conflict(state, conflicts)

    conf_penalty = n_course_conf * 10

    n_exc_courses = n_excess_courses(state , S)

    dist_penalty = n_exc_courses * 2

    return conf_penalty , dist_penalty , conf_penalty + dist_penalty

```

# Neighbor Generation

```python
def generate_neighbors(initial_state , S):

  all_neighbours = []

  for i in range(len(initial_state)):
    for j in range(S):
      if initial_state[i] == j+1 :
        continue

      new_neighbour = initial_state.copy()
      new_neighbour[i] = j+1

      all_neighbours.append(new_neighbour)

  return all_neighbours
```

# Hill Climbing

```python
def hill_climbing(initial_state , conflicts ,S):

  x , y , initial_state_cost = calculate_cost(initial_state , conflicts , S)

  all_neighbours = generate_neighbors(initial_state , S)

  best_neighbour = initial_state
  best_neighbour_cost = initial_state_cost

  for neighbour in all_neighbours:

    x , y , neighbour_cost = calculate_cost(neighbour , conflicts , S)

    if neighbour_cost < best_neighbour_cost:
      best_neighbour = neighbour
      best_neighbour_cost = neighbour_cost

  return best_neighbour , best_neighbour_cost

```

```python
import time as time

def main(N,S,K,conflicts,initial_state):

  start_time = time.time()
  print("========================================")
  print("Algorithm: Hill Climbing")
  print("Problem: Exam Timetable Optimization")
  print("========================================")

  print("Initial Timetable :")

  for i in range(N):
    print("C",(i+1),"-> Slot " ,initial_state[i])

  conflict_cost , dist_cost , total_cost = calculate_cost(initial_state , conflicts , S)
  print("Initial Conflict Cost = ",conflict_cost)
  print("Initial Distribution Cost = ",dist_cost)
  print("Initial Total Cost = ",total_cost)

  i=1

  while(1):

    best_neighbour,best_neighbour_cost = hill_climbing(initial_state , conflicts , S)

    print("----------------------------------------")
    print("Iteration ",i)
    print("Best Neighbor Timetable:")
    for j in range(N):
      print("C",(j+1),"-> Slot " ,best_neighbour[j])

    neighbour_conflict_cost , neighbour_dist_cost , neighbour_total_cost = calculate_cost(best_neighbour , conflicts , S)
    print("Conflict Cost = ",neighbour_conflict_cost)
    print("Distribution Cost = ",neighbour_dist_cost)
    print("Total Cost = ",neighbour_total_cost)

    if best_neighbour_cost >= total_cost:
      break

    initial_state = best_neighbour
    total_cost = best_neighbour_cost
    i=i+1

  print("========================================")
  print("Final Timetable :")

  for j in range(N):
    print("C",(j+1),"-> Slot " ,initial_state[j])

  conflict_cost , dist_cost , total_cost = calculate_cost(initial_state , conflicts , S)
  print("Final Conflict Cost = ",conflict_cost)
  print("Final Distribution Cost = ",dist_cost)
  print("Final Total Cost = ",total_cost)

  print("Interations = ",i)
  print("Total Neighboring States Evaluated = ",(18*i))
  end_time = time.time()
  print("Execution Time = ",end_time-start_time," seconds")

  if total_cost == 0:
    print("Termination Reason: Global Optimum (Ideal Solution)")
  else:
    print("Termination Reason: Local Optimum")
  print("========================================")
```
