```python
graph = {
    0 : "S..#.",
    1 : "#..#.",
    2 : ".....",
    3 : ".##..",
    4 : "....G"
}
R = len(graph)
C = len(graph[0])
temp = [0 for _ in range(C)]
dirn = [(-1,0),(0,1),(1,0),(0,-1)]
dirnPath = ["UP","RIGHT","DOWN","LEFT"]
```

```python
def successor(row : int, col : int, visted : list[list[int]]) -> list[tuple[int,int,str]]:
    successors = []
    i = 0
    for dr, dc in dirn:
        nr, nc = row + dr, col + dc
        if (
            nr >= 0 and nr < R
            and nc >= 0 and nc < C
            and visted[nr][nc] == 0
            and graph[nr][nc] != "#"
        ): 
            successors.append((nr,nc,dirnPath[i]))
        i += 1
    return successors
```

```python
def start_postion() -> tuple[int,int]:
    for i in range(R):
        for j in range(C):
            if graph[i][j] == "S":
                return i, j
    return -1, -1
```

```python
def bfs() -> tuple[list[str],int] | None:
    visted = [temp.copy() for _ in range(R)]
    r, c = start_postion()
    nodesExp = 0
    queue = []
    queue.append((r,c,[]))
    while(len(queue) > 0):
        r, c, path = queue.pop(0)
        nodesExp += 1
        visted[r][c] = 1
        if graph[r][c] == "G":
            return path, nodesExp
        for nr, nc, ndirn in successor(r,c,visted):
            queue.append((nr,nc,path + [ndirn]))
    return None
```

```python
bfs()
```

```python
visted = [temp.copy() for _ in range(R)]
nodesExp = [0]
r, c = start_postion()

def dfs(visted : list[list[int]], nodesExp : list[int], r : int, c : int) -> tuple[bool,list[str] | None]:
    visted[r][c] = 1
    nodesExp[0] += 1
    if graph[r][c] == "G":
        return True, []
    for nr, nc, ndirn in successor(r,c,visted):
        isPos, path = dfs(visted,nodesExp,nr,nc)
        if isPos and path is not None:
            return True, [ndirn] + path
    return False, None
```

```python
dfs(visted, nodesExp, r, c)
```
